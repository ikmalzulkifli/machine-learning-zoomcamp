```python
X_df = asian_cars[['vehicle_weight', 'model_year']]
```

```python
X_df_subset = X_df.head(7)
```

```python
X = X_df_subset.to_numpy()
print(f"1. Final Matrix X (Shape: {X.shape}):\n{X}\n")
```

```python
XT = X.T
XTX = XT @ X
print(f"2. Matrix XTX:\n{XTX}\n")
```

```python
import numpy as np
XTX_inv = np.linalg.inv(XTX)
print(f"3. Inverse of XTX (XTX_inv):\n{XTX_inv}\n")
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ XT @ y
print(f"5. Resulting vector w (Least Squares Weights):\n{w}\n")
```

```python
sum_of_w = w.sum()
```
